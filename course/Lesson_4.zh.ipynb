### 4. 代码中的注释

你可能想知道，上面单元格中以 `#` 符号开头的文字是什么意思。这是注释行，它告诉 Python：可以忽略这一行中 `#` 后面的所有内容。注释用于帮助其他人理解你的代码在做什么。

试着按 `Shift` + `Enter` 运行下一个单元格。你觉得会发生什么？

```python
# THIS IS A COMMENT - PYTHON WILL IGNORE THIS LINE
```

上面的单元格和之前的代码一样正常运行：你会看到单元格左侧出现 `*`，随后变成一个数字。不过，这次屏幕上没有出现任何输出。

注释行也是有效的 Python 代码，它告诉 Python 忽略后面的内容。这里，Python 严格执行了这条指令：运行代码、忽略注释文字。由于这段代码没有需要输出的内容，看起来就像“什么也没发生”。

点击下方单元格，按 `Shift` + `Enter` 运行代码。先想一想会发生什么，再看看你的判断是否正确。
