## 结合字符串与计算或数据：f-string

如果要把华氏温度换算为摄氏温度，并显示结果，可以这样写：

```python
print(((75 - 32) * 5 / 9))
```

如果希望在一段文字中显示换算结果，你可能会尝试这样写：

```python
print("The temperature 75F in degrees Celsius is ((75 - 32) * 5 / 9)C")
```

但这样只会原样输出计算公式，不会真正完成温度换算。

要在文字中输出计算结果，可以使用 Python 的 **f-string（格式化字符串）**，如下所示：

```python
print(f"The temperature 75F in degrees celsius is {(75 - 32) * 5 / 9}C")
```

在表示字符串的引号前加上 `f`（formatted 的缩写），再把计算放进花括号 `{(75 - 32) * 5 / 9}` 中。`f` 告诉 Python：需要计算字符串内花括号中的表达式，并把结果放到相应位置。

## 控制 f-string 中数据的显示方式

再看一个例子。可以用下面的代码输出 Isabel 的年龄：

```python
print("Isabel is 28 years old.")
```

在一些地方，人们会把人的年龄换算成“狗龄”。本课使用一个简化的练习规则：狗龄等于人的年龄除以 7。如果想计算 Isabel 的狗龄并清楚地显示出来，下面这种普通字符串的写法不会进行计算。

```python
print("Isabel is 28/7 dog years old.")
```

应该像温度换算的例子一样，使用 f-string，把狗龄的计算放进花括号中。

```python
print(f"Isabel is {28/7} dog years old.")
```

上一个单元格已经能显示计算结果，但结果带有小数。如果希望像通常描述年龄那样，只显示整数，可以请聊天机器人帮忙。你可以使用下面的提示词，也可以参考它自行提问：

<p style="background-color:#F5C780; padding:15px">🤖 <b>向聊天机器人提问</b>：修改这段代码，让结果不显示小数部分：<code>print(f"Isabel's dog age is {28/7}.")</code></p>

你也可以直接在本页编辑代码完成练习。

使用上面的提示词或类似问法，可能会得到下面这样的代码。

```python
print(f"Isabel is {28/7:.0f} dog years old.")
```

在 `f"{28/7:.0f}"` 中，`:.0f` 表示把 `28/7` 的结果按浮点数格式显示，并保留 0 位小数。你暂时不必记住全部细节：它会按 Python 的舍入规则显示整数结果，而不是直接删除小数部分。

**你刚刚学习了如何借助 AI 修改代码！** 先提供一段能显示小数的代码，再要求它改为不显示小数。这就是用清晰的要求指导 AI 调整程序行为。

## 多行 f-string

到目前为止，你使用的字符串都比较短。对于较长的文字，多行字符串在代码编辑器里更容易阅读。

下面的多行 f-string 介绍了美国食谱中常见的非公制计量单位，并计算 8 液体盎司牛奶对应多少毫升，以及 100 毫升水对应多少液体盎司。

```python
print(f"""
    Most countries use the metric system for recipe measurement, 
    but American bakers use a different system. For example, they use 
    fluid ounces to measure liquids instead of milliliters (ml).
    
    So you need to convert recipe units to your local measuring system!
    
    For example, 8 fluid ounces of milk is {8 * 29.5735} ml.
    And 100ml of water is {100 / 29.5735} fluid ounces.
""")
```

多行字符串要使用三引号 `"""`，而不是一对普通双引号 `"`。字符串可以包含多行文字和计算，其他用法与之前基本相同。

你可能已经发现，直接把很多数字和计算放进花括号，会让代码不容易一眼看懂。计算越多，阅读时越容易分心。

下一课将介绍如何让代码更容易阅读，同时更加灵活、方便重复使用。
