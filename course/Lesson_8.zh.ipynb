# 第 8 课：变量

在编程中，**变量**用于存储、处理和操作数据。例如，要把 `28` 存入名为 `age` 的变量，可以这样写：

```python
age = 28
```

接着把 `age` 传给 `print()`，就会输出你赋给这个变量的值。

```python
print(age)
```

`age = 28` 告诉 Python 创建一个名为 `age` 的变量，并用它存储 `28`。已经创建的变量也可以被重新赋值。例如，下面把 `age` 改为 `5`：

```python
age = 5
```

现在输出 `age`，得到的就是 `5`，而不是 `28`。

```python
print(age)
```

## 变量可以存储数字、字符串和其他类型的数据

变量可以存储浮点数、整数、字符串以及其他类型的数据。例如，创建变量 `name` 并赋值为字符串 `"Otto"`，再创建变量 `gnome_height` 并赋值为浮点数 `12.7`。

```python
name = "Otto"
gnome_height = 12.7
```

把变量放进 f-string，可以让输出更清楚。例如，在变量 `age` 的值前加上文字 `"Age: "`，读者就能知道这个数字表示年龄。

写法如下：

```python
print(f"Age: {age}")
```

再用 f-string 输出变量 `name` 和 `gnome_height` 的值：

```python
print(f"Name: {name}")
print(f"Gnome height: {gnome_height}")
```

**注意：变量名区分大小写。** `Gnome_height` 和 `gnome_height` 是两个不同的名字。运行下一个单元格，会看到相应的错误信息。

```python
print(f"Gnome height: {Gnome_height}")
```

## 用变量保存不断变化的值

假设一个游戏的得分从零开始。在下一个单元格中，创建变量 `score`，把它设为 `0`，再显示当前得分。

```python
score = 0 # now score is 0
print(score)
```

得分增加 50 分时，可以在变量原来的值上加上 `50`，再把结果赋给该变量。

```python
score = score + 50 # now score is 0 + 50 which is 50
print(score)
```

同样，得分再增加 100 分时，可以把 `100` 加到 `score` 当前保存的值上，然后更新 `score`。

```python
score = score + 100 # now score is 100 + 50 which is 150
print(score)
```

再获得 300 分时，把 `300` 加到 `score` 当前的值上，再次更新得分。

```python
score = score + 300 # now score is 150 + 300 which is 450
print(score)
```

这样，变量 `score` 最终保存的就是最新得分。

可以像下面这样，用 f-string 输出最终得分：

```python
print(f"Your final score was: {score}") # prints 450
```

## 变量让代码更高效！

回顾前几课中把人的年龄除以 7 来计算“狗龄”的练习。假设 Otto 是 49 岁，可以用 `print(49 / 7)` 显示他的狗龄。

```python
print(49 / 7)
```

为了让读者知道数字表示什么，可以用下面这样的 f-string：

```python
print(f"Otto's age in dog years is {49 / 7}")
```

了解变量之后，也可以先算出 Otto 的狗龄，把结果存入变量 `dog_age`。

```python
dog_age = 49 / 7
```

如果要在同一段文字中多次使用 Otto 的狗龄，变量就很有帮助。不使用变量时，每出现一次，都要再写一遍计算式。

```python
print(f"""Otto's dog age is {49/7}. So a dog that's about
{49/7} would be the same age as Otto. Any dog born about {49/7}
years ago would be in the same stage of life as Otto.""")
```

如果 Otto 又长大一岁，就需要把每一处花括号中的计算都改为 `50/7`。如果改用下面这样包含 `dog_age` 变量的 f-string：

```python
print(f"""Otto's dog age is {dog_age}. So a dog that's about
{dog_age} would be the same age as Otto. Any dog born about {dog_age}
years ago would be in the same stage of life as Otto.""")
```

就只需要用新的狗龄更新 `dog_age`：

```python
dog_age = 50/7
```

然后再次运行之前的 f-string，无须修改它的内容。一个变量可以在多个位置使用，让程序更容易维护和复用。

```python
print(f"""Otto's dog age is {dog_age}. So a dog that's about
{dog_age} would be the same age as Otto. Any dog born about {dog_age}
years ago would be in the same stage of life as Otto.""")
```

还可以把文字中的名字 `"Otto"` 换成花括号中的变量 `name`。

```python
print(f"""{name}'s dog age is {dog_age}. So a dog that's about
{dog_age} would be the same age as {name}. Any dog born about {dog_age}
years ago would be in the same stage of life as {name}.""")
```

这样就可以把 Otto 的名字改为包含姓和名的完整姓名：

```python
name = "Otto Matic"
```

再次运行这个 f-string，所有使用 `name` 的位置都会显示更新后的名字。

```python
print(f"""{name}'s dog age is {dog_age}. So a dog that's about
{dog_age} would be the same age as {name}. Any dog born about {dog_age}
years ago would be in the same stage of life as {name}.""")
```
