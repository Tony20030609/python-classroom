# 第 6 课：Python 中的数据

Python 中有多种数据类型。本课将介绍如何在 Python 中使用文字和数字。

## 字符串

字符串用于存储和处理文字。如下所示，字符串写在引号内，可以包含字母、数字、标点和其他特殊字符。运行下面的单元格，输出这四个不同的字符串。

```python
print("Hello, World")
```

```python
print("My favorite drink is Earl Grey tea.")
```

```python
print("¯\_(ツ)_/¯")
```

```python
print("2.99")
```

## 多行字符串

使用三引号可以创建多行字符串，让文字跨越多行。运行下面的单元格，你会看到第二行开头的空格也会作为字符串中的字符保留下来。

```python
print("""Hello, World!
      It's great to be here!""")
```

如果只用一对普通引号来定义跨行字符串，就会出错。

```python
print("Hello, World!
      It's great to be here!")
```

## `type()` 函数

在 Python 中，可以用 `type()` 函数查看数据类型。运行下一个单元格，查看字符串 `"Andrew"` 的类型。

```python
type("Andrew")
```

Python 返回了 `str`，它是 string（字符串）的缩写。接下来，查看一个多行字符串的类型：

```python
type("""
Numbers, text, and truth,
Strings, ints, and floats in our code,
Data shapes our path
""")
```

结果同样是 `str`，因为它也是字符串。再试试用引号括起来的数字：

```python
type("2.99")
```

虽然看起来像数字，它仍然是字符串。加上引号，就表示让 Python 把它当作文字处理。去掉引号之后会怎样？

```python
type(100)
```

这次返回了 `int`，即 integer（整数）的缩写。整数包括正整数、负整数和零，例如 42、100、-9 和 0。这个数字没有引号，也没有小数点，因此 Python 把它识别为整数。接下来试试带小数点的数字：

```python
type(2.99)
```

`type()` 返回了 `float`，这是用于存储浮点数的数据类型。浮点数可以是正数或负数，并带有小数部分，例如 3.14、2.99 和 -0.003。

### 运算顺序

Python 的运算顺序与通常的算术规则相同：先算括号，再算乘方，然后从左到右计算乘除，最后从左到右计算加减。

因此，如果你想把华氏温度转换为摄氏温度，下面的写法会得到错误结果：

```python
print(75 - 32 * 5 / 9)
```

而下面单元格中的计算顺序是正确的。

```python
print((75 - 32) * 5 / 9)
```
